```python
from datasets import load_dataset
import pandas as pd
```

```python
eng_arc = load_dataset('allenai/ai2_arc', 'ARC-Challenge', split='test')
eng_arc
```

```python
eng_arc.to_pandas()
```

```python
llama_eus_examples = pd.read_json(
    "/gscratch5/users/asalem/cpt-ift/results/icl_benchmarks/meta-llama/Llama-3.1-8B/arc_eu_challenge/5-shot.json/meta-llama__Llama-3.1-8B/samples_arc_eu_challenge_2025-03-25T12-30-30.135362.jsonl",
    lines=True,
)
```

```python
llama_eus_examples['doc_id_arc'] = llama_eus_examples['doc'].apply(lambda x: x['id'])
```

```python
llama_eus_examples.merge(eng_arc.to_pandas(), left_on='doc_id_arc', right_on='id').to_csv("./llama_eus_arc_challenge.csv", index=False)
```

```python
eus_ids = set(llama_eus_examples['doc_id_arc'].unique())
eng_ids = set(eng_arc.to_pandas()['id'].unique())
```
